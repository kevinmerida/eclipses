# Angle de l'inclinaison apparente de l'axe de rotation de la Terre

* $\lambda$ : longitude écliptique du Soleil
* $\epsilon$ : inclinaison de l'axe de rotation de la Terre
* $\delta$ : latitude du point subsolaire
* $\theta$ : inclinaison apparente de l'axe de rotation de la Terre


```python
import sympy as sp
from IPython.display import display, Math

lbd, eps, dlt, tht = sp.symbols('\\lambda, \\epsilon, \\delta, \\theta')
```

Le repère géocentrique équatorial est XYZ avec X point vernal, Z pôle nord céleste

Matrice de rotation pour la longitude écliptique $\lambda$

```python
Rlbd = sp.Matrix([[sp.cos(lbd), -sp.sin(lbd), 0],
                 [sp.sin(lbd), sp.cos(lbd), 0], [0, 0, 1]])
Rlbd
```

Matrice de rotation pour l'inclinaison $\epsilon$

```python
Reps = sp.Matrix([[1, 0, 0], [0, sp.cos(eps), -sp.sin(eps)],
                 [0, sp.sin(eps), sp.cos(eps)]])
Reps
```

Combinaison des deux rotations

```python
Reps@Rlbd
```

Pôle écliptique dans XYZ

```python
v_eclipt = sp.Matrix([[0], [-sp.sin(eps)], [sp.cos(eps)]])
v_eclipt
```

Direction du point subsolaire dans XYZ

```python
x_view = Reps@Rlbd@sp.Matrix([[1], [0], [0]])
x_view
```

Vecteur unitaire y_view orthogonal à x_view et dans le plan équatorial

```python
y_view = sp.Matrix([[0], [0], [1]]).cross(x_view)
y_view = y_view/sp.sqrt(y_view.dot(y_view))
y_view
```

z_view est le produit vectoriel de x_view et y_view

Ce vecteur est dans le plan qui est orthogonal au plan équatorial et qui inclut le point subsolaire.

```python
z_view = x_view.cross(y_view)
z_view
```

Coordonnée du pôle écliptique v_eclipt sur l'axe y_view

```python
v_eclipt.dot(y_view).simplify()
```

Coordonnée du pôle écliptique v_eclipt sur l'axe z_view

```python
v_eclipt.dot(z_view).simplify()
```

Tangente de l'inclinaison apparente du pôle écliptique dans le plan de projection (y_view, z_view)

```python
tant = v_eclipt.dot(y_view)/v_eclipt.dot(z_view)
tant = tant.simplify()
tant
```

Résolution des équations

```python
sol = sp.solve([sp.sin(dlt)-sp.sin(lbd)*sp.sin(eps),
               sp.tan(tht)-sp.cos(lbd)*sp.tan(eps)], [tht, lbd])
```

Relation liant $\lambda$, $\delta$ et $\epsilon$

```python
display(Math('\\lambda = '+sp.latex(sol[1][1])))
```

Valeur de l'angle de rotation à appliquer dans le plan (y_view, z_view), du solstice de juin au solstice de décembre

```python
display(Math('\\theta='+sp.latex(sol[1][0])))
```
